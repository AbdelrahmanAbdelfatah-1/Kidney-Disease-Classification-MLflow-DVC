# What we can do with ConfigBox

```python
from box import ConfigBox
```

```python
d1 = ConfigBox({"a": 1, "b": 2})
```

```python
d1.a
```

```python
from ensure import ensure_annotations
```

```python
@ensure_annotations
def test(a: int, b: int) -> int:
    return a * b
```

```python
f = test(2, 3)
f
```

```python
f = test(2, "3")
f
```
