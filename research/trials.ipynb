# What we can do with "@ensure_annotations"

```python
from ensure import ensure_annotations
```

```python
@ensure_annotations
def test(a: int, b: int) -> int:
    return a * b
```

```python
f = test(2, 3)
f
```

```python
f = test(2, "3")
f
```
